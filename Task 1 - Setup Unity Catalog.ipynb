### Tạo Catalog và các Schema:

```python
%sql
-- Tạo catalog riêng cho project
CREATE CATALOG IF NOT EXISTS nyc_taxi_project
  COMMENT 'NYC Yellow Taxi ETL Pipeline Project';

-- Tạo các schema theo Medallion Architecture
CREATE SCHEMA IF NOT EXISTS nyc_taxi_project.landing
  COMMENT 'Landing zone - raw Parquet files';

CREATE SCHEMA IF NOT EXISTS nyc_taxi_project.bronze
  COMMENT 'Bronze layer - raw ingested data';

CREATE SCHEMA IF NOT EXISTS nyc_taxi_project.silver
  COMMENT 'Silver layer - cleaned data';

CREATE SCHEMA IF NOT EXISTS nyc_taxi_project.gold
  COMMENT 'Gold layer - aggregated data';
```

### Tạo Volume để chứa file Parquet raw

```python
%sql
-- Volume = nơi lưu file thô
CREATE VOLUME IF NOT EXISTS nyc_taxi_project.landing.raw_files
  COMMENT 'Landing zone for NYC Taxi Parquet files';
```

### Tạo Volumn cho checkpoint

```python
%sql
-- Tạo volume riêng để lưu checkpoint
CREATE VOLUME IF NOT EXISTS nyc_taxi_project.landing.checkpoints
  COMMENT 'Checkpoint location for streaming jobs';
```

### Tạo Volumn cho lookup

```python
%sql
-- Tạo volume lookup_files trong Unity Catalog
CREATE VOLUME IF NOT EXISTS nyc_taxi_project.landing.lookup_files
  COMMENT 'Lookup/reference files (CSV, JSON...)';
```
